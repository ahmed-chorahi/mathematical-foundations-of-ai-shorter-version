# Objective & Loss Functions

*📓 Notes · [05 · Optimization](../README.md) · what to minimize*

**📝 In simple words:** A single number saying how bad the model currently is.

**💡 Think of it like this:** A score we want to push down.

**🔢 Small example:** predict `ŷ = 8` when truth is `y = 10`.

**📐 Formula to remember:** squared error `L = (y − ŷ)²`

**✏️ Let me calculate it:** `(10 − 8)² = 4`.

**🤖 Why AI uses it:** without a loss there's nothing to minimize. Details in chapter 07.

````python
y, y_hat = 10, 8
print((y - y_hat) ** 2)   # 4
````

**🧪 Try it myself:** Loss for `y=3`, `ŷ=5`? <details><summary>Solution</summary>`4`.</details>

# Local vs Global Minima

*📓 Notes · [05 · Optimization](../README.md) · what to minimize*

**📝 In simple words:** A local minimum is the lowest point nearby; the global one is the lowest overall.

```text
 loss
  │ \      /\
  │  \    /  \      /
  │   \__/    \____/
  │    ↑local    ↑global
  └──────────────────► w
```

**🤖 Why AI uses it:** in very high dimensions most trouble comes from saddle points and flat regions, not bad local minima. Convex losses (like linear regression's) have only one minimum.

**🧪 Try it myself:** How many minima does `x²` have? <details><summary>Solution</summary>One, the global minimum at 0.</details>

# Gradient Descent

*📓 Notes · [05 · Optimization](../README.md) · gradient descent family*

**📝 In simple words:** Repeatedly step downhill.

**📐 Formula to remember:** `w ← w − α ∇L(w)`

**🔢 Small example:** minimize `L(w) = (w − 3)²`, start `w = 0`, `α = 0.1`.

**✏️ Let me calculate it:**

```text
grad = 2(w − 3)
w₀=0      grad=−6      w₁ = 0 − 0.1(−6) = 0.6
w₁=0.6    grad=−4.8    w₂ = 0.6 + 0.48  = 1.08
w₂=1.08   grad=−3.84   w₃ = 1.464 ...→ 3
```

**🤖 Why AI uses it:** it's how every neural network is trained.

````python
w, lr = 0.0, 0.1
for step in range(30):
    w -= lr * 2 * (w - 3)
print(round(w, 4))   # ≈ 2.9988
````

**🧪 Try it myself:** One step from `w=5` on the same loss with `α=0.25`? <details><summary>Solution</summary>`grad=4`; `w = 5 − 1 = 4`.</details>

# Learning Rate

*📓 Notes · [05 · Optimization](../README.md) · gradient descent family*

**📝 In simple words:** The step size `α`.

**💡 Think of it like this:** Too small = slow. Too large = overshoot and diverge.

**✏️ Let me calculate it (α = 1.1):** `w₁ = 0 − 1.1(−6) = 6.6`, `grad = 7.2`, `w₂ = 6.6 − 7.92 = −1.32`. It bounces further away every step.

**🤖 Why AI uses it:** probably the most important hyperparameter in training.

````python
for lr in (0.01, 0.1, 1.1):
    w = 0.0
    for _ in range(20): w -= lr * 2 * (w - 3)
    print(lr, round(w, 3))     # slow, good, blows up
````

**🧪 Try it myself:** For this loss, what `α` jumps exactly to the minimum in one step? <details><summary>Solution</summary>`α = 0.5` (`w − 0.5·2(w−3) = 3`).</details>

# SGD & Mini-batch Gradient Descent

*📓 Notes · [05 · Optimization](../README.md) · gradient descent family*

**📝 In simple words:** Use a small random chunk of data per step instead of everything.

**💡 Think of it like this:** Polling 32 people tells you roughly what 1 million think, and it's way cheaper.

**📐 Formula to remember:** `∇L ≈ (1/B) Σᵢ∈batch ∇ℓᵢ`

| Type | Batch size | Trait |
|---|---|---|
| Batch GD | all data | smooth, slow |
| SGD | 1 | noisy, fast |
| Mini-batch | 32–512 | the standard compromise |

**🤖 Why AI uses it:** datasets are too big to use in one step; noise even helps escape bad spots.

````python
import numpy as np
rng = np.random.default_rng(0)
X = rng.normal(size=1000); y = 3 * X + rng.normal(0, 0.1, 1000)
w, lr = 0.0, 0.1
for step in range(200):
    idx = rng.integers(0, 1000, 32)               # mini-batch of 32
    grad = np.mean(2 * (w * X[idx] - y[idx]) * X[idx])
    w -= lr * grad
print(round(w, 2))   # ≈ 3
````

**🧪 Try it myself:** 1,000 samples, batch 100: steps per epoch? <details><summary>Solution</summary>10.</details>

# Momentum

*📓 Notes · [05 · Optimization](../README.md) · smarter optimizers*

**📝 In simple words:** Keep part of the previous direction, like a rolling ball.

**📐 Formula to remember:** `v ← βv + ∇L`, `w ← w − αv` (β ≈ 0.9)

**✏️ Let me calculate it:** gradients `−6` then `−4.8`: `v₁ = −6`; `v₂ = 0.9(−6) − 4.8 = −10.2`. The step grows when the direction stays consistent.

**🤖 Why AI uses it:** speeds up along valleys and smooths noisy mini-batch gradients.

````python
w, v = 0.0, 0.0
for _ in range(50):
    g = 2 * (w - 3); v = 0.9 * v + g; w -= 0.05 * v
print(round(w, 3))
````

**🧪 Try it myself:** If `v=2`, `β=0.9`, new gradient `1`: new `v`? <details><summary>Solution</summary>`1.8 + 1 = 2.8`.</details>

# Adam

*📓 Notes · [05 · Optimization](../README.md) · smarter optimizers*

**📝 In simple words:** Momentum plus a per-weight adaptive step size.

**📐 Formula to remember:**

```text
m ← β₁m + (1−β₁)g           (average of gradients)
v ← β₂v + (1−β₂)g²          (average of squared gradients)
m̂ = m/(1−β₁ᵗ),  v̂ = v/(1−β₂ᵗ)
w ← w − α · m̂ / (√v̂ + ε)
```

**✏️ Let me calculate it (first step, g = −6):** `m = −0.6`, `v = 0.036`; `m̂ = −6`, `v̂ = 36`; step `= α·(−6)/6 = −α`. Each weight moves about `α` regardless of gradient scale.

**🤖 Why AI uses it:** the default optimizer for Transformers/LLMs (AdamW).

````python
w, m, v = 0.0, 0.0, 0.0
b1, b2, lr, eps = 0.9, 0.999, 0.1, 1e-8
for t in range(1, 201):
    g = 2 * (w - 3)
    m = b1*m + (1-b1)*g; v = b2*v + (1-b2)*g*g
    mh, vh = m/(1-b1**t), v/(1-b2**t)
    w -= lr * mh / (np.sqrt(vh) + eps)
print(round(w, 2))   # ≈ 3
````

**🧪 Try it myself:** Why divide by `√v̂`? <details><summary>Solution</summary>Weights with big gradients get smaller steps and weights with tiny gradients get larger ones.</details>

# Constrained Optimization

*📓 Notes · [05 · Optimization](../README.md) · constraints*

**📝 In simple words:** Minimize a loss while obeying a rule.

**🔢 Small example:** minimize `x² + y²` subject to `x + y = 1`.

**📐 Formula to remember (Lagrange):** `ℒ = f(x,y) + λ(x + y − 1)`; set all partial derivatives to 0.

**✏️ Let me calculate it:** `2x + λ = 0`, `2y + λ = 0` → `x = y`; `x + y = 1` → `x = y = 0.5`, value `0.5`.

**🤖 Why AI uses it:** L2/L1 regularization is a soft constraint on weight size; SVMs are constrained problems; softmax outputs must sum to 1.

````python
x = np.linspace(0, 1, 101); y = 1 - x
print(x[np.argmin(x**2 + y**2)])   # 0.5
````

**🧪 Try it myself:** Minimize `x² + y²` with `x + y = 4`. <details><summary>Solution</summary>`x = y = 2`, value 8.</details>
