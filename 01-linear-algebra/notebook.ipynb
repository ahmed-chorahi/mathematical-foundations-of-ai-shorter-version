# Scalars

*📓 Notes · [01 · Linear Algebra](../README.md) · building blocks*

**📝 In simple words:** A single number. `3`, `−0.5`.

**💡 Think of it like this:** One measurement, like temperature.

**🔢 Small example:** learning rate `α = 0.01`.

**📐 Formula to remember:** `x ∈ ℝ`

**✏️ Let me calculate it:** scaling a vector by 2: `2 · [1, 3] = [2, 6]`.

**🤖 Why AI uses it:** learning rates, biases, loss values are scalars.

````python
import numpy as np
print(2 * np.array([1, 3]))   # [2 6]
````

# Vectors

*📓 Notes · [01 · Linear Algebra](../README.md) · building blocks*

**📝 In simple words:** An ordered list of numbers.

**💡 Think of it like this:** An arrow from the origin, or a row in a spreadsheet. A house = `[size, rooms, age]`.

**🔢 Small example:** `x = [2, 3]`, `y = [1, 4]`

**📐 Formula to remember:** `x + y = [x₁+y₁, x₂+y₂]`

**✏️ Let me calculate it:** `[2,3] + [1,4] = [3, 7]`

```text
 y
 4|     • y
 3| • x
  |
  +--------- x
```

**🤖 Why AI uses it:** a data point, a word embedding, and the weights of a neuron are all vectors.

````python
x = np.array([2, 3]); y = np.array([1, 4])
print(x + y)   # [3 7]
````

**🧪 Try it myself:** `[1,2,3] + [4,0,−1]`? <details><summary>Solution</summary>`[5, 2, 2]`</details>

# Matrices

*📓 Notes · [01 · Linear Algebra](../README.md) · building blocks*

**📝 In simple words:** A grid of numbers (rows × columns).

**💡 Think of it like this:** A stack of vectors. A dataset: each row is one sample.

**🔢 Small example:** `A = [[1,2],[3,4]]` is 2×2.

**📐 Formula to remember:** `A ∈ ℝ^(m×n)` has m rows, n columns.

**✏️ Let me calculate it:** transpose flips rows and columns: `Aᵀ = [[1,3],[2,4]]`.

**🤖 Why AI uses it:** weights of a layer are a matrix; a batch of data is a matrix.

````python
A = np.array([[1, 2], [3, 4]])
print(A.shape, A.T)
````

# Matrix Operations

*📓 Notes · [01 · Linear Algebra](../README.md) · operations and distance*

**📝 In simple words:** Add, scale, and multiply matrices.

**💡 Think of it like this:** Matrix × vector = each row takes a weighted sum of the vector. (It's many dot products at once.)

**🔢 Small example:** `A = [[1,2],[3,4]]`, `v = [5,6]`

**📐 Formula to remember:** `(Av)ᵢ = Σⱼ Aᵢⱼ vⱼ`. For `A(m×n) · B(n×p)` the result is `m×p`.

**✏️ Let me calculate it:** row 1: `1·5 + 2·6 = 17`; row 2: `3·5 + 4·6 = 39` → `[17, 39]`

**🤖 Why AI uses it:** `y = Wx + b` is a neural-network layer.

````python
A = np.array([[1, 2], [3, 4]]); v = np.array([5, 6])
print(A @ v)   # [17 39]
B = np.array([[0, 1], [1, 0]])
print(A @ B)   # swaps columns of A
````

**🧪 Try it myself:** `[[2,0],[0,3]] · [4,5]`? <details><summary>Solution</summary>`[8, 15]`. It just scales each coordinate.</details>

# Dot Product

*📓 Notes · [01 · Linear Algebra](../README.md) · operations and distance*

**📝 In simple words:** Multiply matching entries and add.

**💡 Think of it like this:** Measures how much two vectors point the same way. Big positive = aligned, 0 = perpendicular, negative = opposite.

**🔢 Small example:** `a = [1,2,3]`, `b = [4,5,6]`

**📐 Formula to remember:** `a·b = Σ aᵢbᵢ = ‖a‖‖b‖ cos θ`

**✏️ Let me calculate it:** `1·4 + 2·5 + 3·6 = 4 + 10 + 18 = 32`

**🤖 Why AI uses it:** a neuron's weighted sum, attention scores, and cosine similarity are all dot products.

````python
a = np.array([1, 2, 3]); b = np.array([4, 5, 6])
print(a @ b)   # 32
````

**🧪 Try it myself:** `[2,−1]·[1,2]`? <details><summary>Solution</summary>`2 − 2 = 0`, so they are perpendicular.</details>

# Norms

*📓 Notes · [01 · Linear Algebra](../README.md) · operations and distance*

**📝 In simple words:** The length (size) of a vector.

**💡 Think of it like this:** How far the arrow reaches.

**🔢 Small example:** `x = [3, 4]`

**📐 Formula to remember:** `L2: ‖x‖ = √(Σxᵢ²)`, `L1: Σ|xᵢ|`

**✏️ Let me calculate it:** L2 = `√(9+16) = 5`; L1 = `3+4 = 7`.

**🤖 Why AI uses it:** regularization penalizes big weights (L1/L2); we normalize vectors to unit length.

````python
x = np.array([3, 4])
print(np.linalg.norm(x), np.linalg.norm(x, 1))   # 5.0 7.0
````

**🧪 Try it myself:** L2 norm of `[1, 2, 2]`? <details><summary>Solution</summary>`√(1+4+4) = 3`.</details>

# Distance

*📓 Notes · [01 · Linear Algebra](../README.md) · operations and distance*

**📝 In simple words:** How far apart two points are: the norm of their difference.

**🔢 Small example:** `a = [1,1]`, `b = [4,5]`

**📐 Formula to remember:** `d(a,b) = ‖a − b‖`

**✏️ Let me calculate it:** `a−b = [−3,−4]` → `√(9+16) = 5`.

**Cosine similarity** (angle-based): `cos θ = a·b / (‖a‖‖b‖)`. For `[1,0]` and `[1,1]`: `1 / (1·1.414) = 0.707`.

**🤖 Why AI uses it:** k-NN, clustering, and embedding search ("find similar") all rely on distance or cosine similarity.

````python
a = np.array([1, 1]); b = np.array([4, 5])
print(np.linalg.norm(a - b))
u = np.array([1, 0]); w = np.array([1, 1])
print(u @ w / (np.linalg.norm(u) * np.linalg.norm(w)))   # 0.707
````

**🧪 Try it myself:** Distance between `[0,0]` and `[6,8]`? <details><summary>Solution</summary>10.</details>

# Linear Transformations

*📓 Notes · [01 · Linear Algebra](../README.md) · transformations and equations*

**📝 In simple words:** A matrix acting as a "move/stretch/rotate everything" machine.

**💡 Think of it like this:** Grid lines stay straight and the origin stays put.

**🔢 Small example:** rotate 90°: `R = [[0,−1],[1,0]]`, apply to `[1, 0]`.

**📐 Formula to remember:** `T(x) = Ax`

**✏️ Let me calculate it:** `[0·1 + (−1)·0, 1·1 + 0·0] = [0, 1]`. The x-axis arrow turned into the y-axis arrow.

**🤖 Why AI uses it:** each layer transforms the data into a new space where the task becomes easier.

````python
R = np.array([[0, -1], [1, 0]])
print(R @ np.array([1, 0]))   # [0 1]
````

**🧪 Try it myself:** What does `[[2,0],[0,1]]` do to `[1,1]`? <details><summary>Solution</summary>`[2,1]`, so it stretches x by 2.</details>

# Systems of Equations

*📓 Notes · [01 · Linear Algebra](../README.md) · transformations and equations*

**📝 In simple words:** Several equations solved together: `Ax = b`.

**🔢 Small example:** `x + y = 3`, `x − y = 1`

**📐 Formula to remember:** `x = A⁻¹ b` (when A is invertible)

**✏️ Let me calculate it:** add them: `2x = 4`, so `x = 2`, `y = 1`.

**🤖 Why AI uses it:** linear regression has a closed-form solution that solves exactly such a system (the "normal equation").

````python
A = np.array([[1, 1], [1, -1]]); b = np.array([3, 1])
print(np.linalg.solve(A, b))   # [2. 1.]
````

**🧪 Try it myself:** Solve `2x + y = 5`, `x − y = 1`. <details><summary>Solution</summary>Add: `3x = 6`, `x = 2`, `y = 1`.</details>

# Eigenvalues & Eigenvectors

*📓 Notes · [01 · Linear Algebra](../README.md) · special directions*

**📝 In simple words:** Special directions that a matrix only stretches, never turns.

**💡 Think of it like this:** Most vectors get knocked off course by a matrix. Eigenvectors stay on their line; the eigenvalue says how much they stretch.

**🔢 Small example:** `A = [[2,0],[0,3]]`

**📐 Formula to remember:** `Av = λv`

**✏️ Let me calculate it:** `A·[1,0] = [2,0] = 2·[1,0]` → λ = 2. `A·[0,1] = [0,3] = 3·[0,1]` → λ = 3.

**🤖 Why AI uses it:** PCA finds the eigenvectors of the covariance matrix: directions of biggest variance.

````python
A = np.array([[2, 0], [0, 3]])
vals, vecs = np.linalg.eig(A)
print(vals)   # [2. 3.]
````

**🧪 Try it myself:** Eigenvalues of `[[4,0],[0,1]]`? <details><summary>Solution</summary>4 and 1 (diagonal entries).</details>

# Projections

*📓 Notes · [01 · Linear Algebra](../README.md) · special directions*

**📝 In simple words:** The "shadow" of one vector on another.

**🔢 Small example:** project `a = [3,4]` onto `b = [1,0]`.

**📐 Formula to remember:** `proj_b(a) = (a·b / b·b) b`

**✏️ Let me calculate it:** `a·b = 3`, `b·b = 1` → `3·[1,0] = [3,0]`.

**🤖 Why AI uses it:** least-squares regression projects `y` onto the space of the features; PCA projects data onto top directions.

````python
a = np.array([3, 4]); b = np.array([1, 0])
print((a @ b) / (b @ b) * b)   # [3. 0.]
````

**🧪 Try it myself:** Project `[2,2]` onto `[0,1]`. <details><summary>Solution</summary>`(2/1)·[0,1] = [0, 2]`.</details>

# Rank

*📓 Notes · [01 · Linear Algebra](../README.md) · special directions*

**📝 In simple words:** The number of truly independent rows/columns.

**💡 Think of it like this:** If one row is just a multiple of another, it adds no new information.

**🔢 Small example:** `[[1,2],[2,4]]`. Row 2 = 2 × row 1.

**📐 Formula to remember:** `rank(A) = dim(column space)`

**✏️ Let me calculate it:** only one independent row → rank 1 (not invertible).

**🤖 Why AI uses it:** redundant (correlated) features give low rank; low-rank structure is behind SVD, compression, and LoRA.

````python
print(np.linalg.matrix_rank(np.array([[1, 2], [2, 4]])))   # 1
````

**🧪 Try it myself:** Rank of `[[1,0],[0,1]]`? <details><summary>Solution</summary>2.</details>

# Orthogonality

*📓 Notes · [01 · Linear Algebra](../README.md) · special directions*

**📝 In simple words:** Perpendicular: dot product = 0.

**🔢 Small example:** `[1,1]` and `[1,−1]`: `1 − 1 = 0`.

**📐 Formula to remember:** `a ⊥ b ⇔ a·b = 0`

**🤖 Why AI uses it:** orthogonal directions carry independent information (PCA components are orthogonal); orthogonal initialization helps training stability.

````python
print(np.array([1, 1]) @ np.array([1, -1]))   # 0
````

**🧪 Try it myself:** Is `[2,3]` ⊥ `[3,−2]`? <details><summary>Solution</summary>`6 − 6 = 0`. Yes.</details>

# PCA Intuition

*📓 Notes · [01 · Linear Algebra](../README.md) · decompositions*

**📝 In simple words:** Rotate the data to find the directions with the most spread, keep the top ones.

**💡 Think of it like this:** A tilted cloud of points is mostly "long" in one direction. Describe each point by its position along that line.

```text
   •  • •
  • • • •        →   keep the long axis (PC1),
 • • • •             drop the thin one (PC2)
```

**✏️ Let me calculate it:** (1) center the data (2) covariance matrix (3) eigenvectors (4) project on top-k.

**🤖 Why AI uses it:** compress features, visualize high-dimensional data, remove noise.

````python
import numpy as np
rng = np.random.default_rng(0)
x = rng.normal(size=200); X = np.c_[x, 0.5*x + 0.1*rng.normal(size=200)]
Xc = X - X.mean(axis=0)
cov = Xc.T @ Xc / (len(Xc) - 1)
vals, vecs = np.linalg.eigh(cov)           # ascending order
print("variance per direction:", vals)      # one big, one tiny
pc1 = vecs[:, -1]; Z = Xc @ pc1             # 2D → 1D
print(Z.shape)
````

**🧪 Try it myself:** If eigenvalues are `[9, 1]`, how much variance does PC1 explain? <details><summary>Solution</summary>`9 / (9+1) = 90%`.</details>

# SVD Intuition

*📓 Notes · [01 · Linear Algebra](../README.md) · decompositions*

**📝 In simple words:** Any matrix = rotate → stretch → rotate: `A = U Σ Vᵀ`.

**💡 Think of it like this:** It finds the most important "layers" of a matrix. Keep the largest singular values for a good approximation.

**📐 Formula to remember:** `A = Σᵢ σᵢ uᵢ vᵢᵀ` with `σ₁ ≥ σ₂ ≥ …`

**✏️ Let me calculate it:** keep only `σ₁` → rank-1 approximation of A.

**🤖 Why AI uses it:** image compression, recommender systems, LoRA (low-rank weight updates), and PCA is SVD in disguise.

````python
A = np.array([[3., 0.], [0., 1.]])
U, S, Vt = np.linalg.svd(A)
print(S)                                    # [3. 1.]
A1 = S[0] * np.outer(U[:, 0], Vt[0])        # rank-1 approximation
print(A1)
````

**🧪 Try it myself:** What's the rank-1 approximation error idea if `S = [10, 0.1]`? <details><summary>Solution</summary>Tiny, since the second layer barely contributes, so rank 1 keeps almost everything.</details>
