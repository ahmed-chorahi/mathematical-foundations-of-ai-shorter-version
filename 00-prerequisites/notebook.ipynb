# Algebra & Equations

*📓 Notes · [00 · Prerequisites](../README.md) · algebra and functions*

**📝 In simple words:** Using letters for unknown numbers and solving for them.

**💡 Think of it like this:** An equation is a balanced scale. Do the same thing to both sides.

**🔢 Small example:** `3x + 2 = 11`

**📐 Formula to remember:** `ax + b = c  →  x = (c − b) / a`

**✏️ Let me calculate it:** `3x = 11 − 2 = 9` → `x = 9 / 3 = 3`

**🤖 Why AI uses it:** A model is just an equation with unknown numbers (weights). Training = solving for them.

````python
a, b, c = 3, 2, 11
print((c - b) / a)   # 3.0
````

**🧪 Try it myself:** Solve `5x − 4 = 21`.
<details><summary>Solution</summary>`5x = 25`, so `x = 5`.</details>

# Functions

*📓 Notes · [00 · Prerequisites](../README.md) · algebra and functions*

**📝 In simple words:** A machine: number in, number out. `f(x)`.

**💡 Think of it like this:** A neuron is a function. A whole network is functions chained together.

**🔢 Small example:** `f(x) = 2x + 1`

**📐 Formula to remember:** `y = f(x)`; composition: `g(f(x))` means "do f, then g".

**✏️ Let me calculate it:** `f(3) = 2·3 + 1 = 7`. If `g(x) = x²`, then `g(f(3)) = 7² = 49`.

**🤖 Why AI uses it:** Layers are composed functions. The chain rule (chapter 02) exists because of composition.

````python
f = lambda x: 2*x + 1
g = lambda x: x**2
print(g(f(3)))   # 49
````

**🧪 Try it myself:** With `f(x)=x+3` and `g(x)=2x`, find `g(f(1))`.
<details><summary>Solution</summary>`f(1)=4`, `g(4)=8`.</details>

# Exponents & Logarithms

*📓 Notes · [00 · Prerequisites](../README.md) · exponents logs and sums*

**📝 In simple words:** `2³` means 2·2·2. A logarithm asks the reverse: "what power gives me this?"

**💡 Think of it like this:** Exponent = repeated multiplying. Log = counting how many multiplications. Logs turn products into sums.

**🔢 Small example:** `2³ = 8`, so `log₂(8) = 3`.

**📐 Formula to remember:**

```text
log(a·b) = log(a) + log(b)
log(a/b) = log(a) − log(b)
log(aⁿ)  = n·log(a)
e^(ln x) = x
```

**✏️ Let me calculate it:** `log(0.5 · 0.5 · 0.5) = 3·log(0.5) = 3 × (−0.693) = −2.079`

**🤖 Why AI uses it:** Probabilities get multiplied and become tiny. Logs make them sums and keep numbers stable. Softmax uses `e^x`; cross-entropy uses `log`.

````python
import numpy as np
print(np.log(0.5**3), 3*np.log(0.5))   # same
print(np.exp(np.log(5)))               # 5.0
````

**🧪 Try it myself:** Compute `log₂(32)`.
<details><summary>Solution</summary>`2⁵ = 32`, so the answer is 5.</details>

# Summation Notation (Σ)

*📓 Notes · [00 · Prerequisites](../README.md) · exponents logs and sums*

**📝 In simple words:** Short way to write "add these up".

**💡 Think of it like this:** A `for` loop that keeps a running total.

**🔢 Small example:** `Σᵢ₌₁³ xᵢ` with `x = [2, 4, 6]`

**📐 Formula to remember:** `Σᵢ₌₁ⁿ xᵢ = x₁ + x₂ + … + xₙ`

**✏️ Let me calculate it:** `2 + 4 + 6 = 12`. The mean is `(1/n) Σ xᵢ = 12/3 = 4`.

**🤖 Why AI uses it:** Dot products, losses (average over data), and expectations are all sums.

````python
x = [2, 4, 6]
total = 0
for v in x: total += v
print(total, total/len(x))   # 12 4.0
````

**🧪 Try it myself:** Compute `Σ xᵢ²` for `x = [1, 2, 3]`.
<details><summary>Solution</summary>`1 + 4 + 9 = 14`.</details>

# Coordinate Systems & Basic Geometry

*📓 Notes · [00 · Prerequisites](../README.md) · geometry and coordinates*

**📝 In simple words:** A way to describe position with numbers; and distance/angle between positions.

**💡 Think of it like this:** `(3, 4)` = 3 steps right, 4 steps up. Distance from origin is the hypotenuse.

**📐 Formula to remember:** `distance = √(x² + y²)` (Pythagoras). Angle from `cos θ = adjacent / hypotenuse`.

```text
 y
 4 |        • (3,4)
   |      /
   |    / 5
   |  /
   +-------- x
          3
```

**✏️ Let me calculate it:** `√(3² + 4²) = √25 = 5`

**🤖 Why AI uses it:** Data points are points in space. "Similar" means "close" or "same direction". 2D intuition carries to 1000D.

````python
import numpy as np
print(np.sqrt(3**2 + 4**2))   # 5.0
````

**🧪 Try it myself:** Distance from origin to `(5, 12)`?
<details><summary>Solution</summary>`√(25+144) = √169 = 13`.</details>
