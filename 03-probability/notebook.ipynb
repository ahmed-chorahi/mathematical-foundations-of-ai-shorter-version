# Random Variables

*📓 Notes · [03 · Probability](../README.md) · basics*

**📝 In simple words:** A variable whose value comes from a random process.

**💡 Think of it like this:** `X` = the number on a die before you roll.

**🔢 Small example:** `X ∈ {1,…,6}`, each with probability `1/6`.

**📐 Formula to remember:** `P(X = x)`; probabilities are `≥ 0` and sum to 1.

**✏️ Let me calculate it:** `P(X=4) = 1/6 ≈ 0.167`; `P(X even) = 3/6 = 0.5`.

**🤖 Why AI uses it:** a model's output ("cat" / "dog") or the next word is a random variable.

````python
import numpy as np
rng = np.random.default_rng(0)
rolls = rng.integers(1, 7, size=10000)
print((rolls % 2 == 0).mean())   # ≈ 0.5
````

**🧪 Try it myself:** Probability of rolling `> 4`? <details><summary>Solution</summary>`2/6 = 1/3`.</details>

# Probability Rules

*📓 Notes · [03 · Probability](../README.md) · basics*

**📝 In simple words:** How to combine probabilities.

**📐 Formula to remember:**

```text
P(not A)    = 1 − P(A)
P(A or B)   = P(A) + P(B) − P(A and B)
P(A and B)  = P(A) · P(B)        (if independent)
```

**🔢 Small example:** two fair coins: `P(both heads) = 0.5 · 0.5 = 0.25`.

**✏️ Let me calculate it:** `P(at least one head) = 1 − P(no heads) = 1 − 0.25 = 0.75`.

**🤖 Why AI uses it:** a sentence's probability is a product of word probabilities; "at least one" tricks show up in evaluation (pass@k).

````python
print(1 - 0.5**2)   # 0.75
````

**🧪 Try it myself:** P(no six in 3 die rolls)? <details><summary>Solution</summary>`(5/6)³ ≈ 0.579`.</details>

# Conditional Probability

*📓 Notes · [03 · Probability](../README.md) · conditional and bayes*

**📝 In simple words:** Probability of A *given* that B happened.

**💡 Think of it like this:** Shrink the world to only cases where B is true.

**🔢 Small example:** 100 emails: 40 spam. 30 spam emails contain "free"; 10 non-spam contain "free".

**📐 Formula to remember:** `P(A|B) = P(A and B) / P(B)`

**✏️ Let me calculate it:** `P(spam | "free") = 30 / (30+10) = 0.75`.

**🤖 Why AI uses it:** a language model is literally `P(next word | previous words)`.

````python
print(30 / (30 + 10))   # 0.75
````

**🧪 Try it myself:** If 20 of the 60 non-spam emails contain "hello" and 10 of the 40 spam do, what's `P(spam | "hello")`? <details><summary>Solution</summary>`10 / 30 = 1/3`.</details>

# Bayes' Theorem

*📓 Notes · [03 · Probability](../README.md) · conditional and bayes*

**📝 In simple words:** Flip a conditional: from `P(B|A)` to `P(A|B)`.

**💡 Think of it like this:** Update your belief after seeing evidence.

**🔢 Small example:** disease affects 1%. A test catches 90% of cases (sensitivity) and falsely flags 5% of healthy people.

**📐 Formula to remember:** `P(A|B) = P(B|A) P(A) / P(B)`

**✏️ Let me calculate it:**

```text
P(pos) = 0.9·0.01 + 0.05·0.99 = 0.009 + 0.0495 = 0.0585
P(disease | pos) = 0.009 / 0.0585 ≈ 0.154
```

Only 15%: the disease is rare, so most positives are false alarms.

**🤖 Why AI uses it:** Naive Bayes classifiers, probabilistic reasoning, and the "prior × likelihood" view of training (MLE vs MAP).

````python
p_d, sens, fpr = 0.01, 0.9, 0.05
p_pos = sens*p_d + fpr*(1 - p_d)
print(sens*p_d / p_pos)   # ≈ 0.154
````

**🧪 Try it myself:** Same test, but disease prevalence 10%? <details><summary>Solution</summary>`P(pos)=0.09+0.045=0.135`; `0.09/0.135 ≈ 0.667`.</details>

# Expected Value

*📓 Notes · [03 · Probability](../README.md) · expectation variance covariance*

**📝 In simple words:** The long-run average outcome.

**🔢 Small example:** fair die.

**📐 Formula to remember:** `𝔼[X] = Σ x · P(x)`

**✏️ Let me calculate it:** `(1+2+3+4+5+6)/6 = 3.5`.

**🤖 Why AI uses it:** the training loss is an *expected* loss over the data; cross-entropy is an expectation.

````python
x = np.arange(1, 7); p = np.full(6, 1/6)
print((x * p).sum())   # 3.5
````

**🧪 Try it myself:** Win $10 with prob 0.2, else lose $2. `𝔼`? <details><summary>Solution</summary>`10(0.2) − 2(0.8) = 0.4`.</details>

# Variance

*📓 Notes · [03 · Probability](../README.md) · expectation variance covariance*

**📝 In simple words:** How spread out the outcomes are around the mean.

**📐 Formula to remember:** `Var(X) = 𝔼[(X − μ)²] = 𝔼[X²] − μ²`

**✏️ Let me calculate it (die):** `𝔼[X²] = (1+4+9+16+25+36)/6 = 15.17`; `Var = 15.17 − 3.5² = 2.92`.

**🤖 Why AI uses it:** noisy gradients, weight initialization scales, and the bias-variance tradeoff.

````python
print((x**2 * p).sum() - 3.5**2)   # ≈ 2.917
````

**🧪 Try it myself:** Variance of a coin (heads=1, tails=0)? <details><summary>Solution</summary>`μ=0.5`; `0.5(0.25)+0.5(0.25) = 0.25`.</details>

# Covariance

*📓 Notes · [03 · Probability](../README.md) · expectation variance covariance*

**📝 In simple words:** Whether two variables move together.

**📐 Formula to remember:** `Cov(X,Y) = 𝔼[(X−μₓ)(Y−μᵧ)]`

**🔢 Small example:** `X=[1,2,3]`, `Y=[2,4,6]`

**✏️ Let me calculate it:** means 2 and 4; products `(−1)(−2)+0+(1)(2) = 4`; divide by `n−1=2` → `2`. Positive → they rise together.

**🤖 Why AI uses it:** the covariance matrix is what PCA decomposes (01).

````python
X, Y = np.array([1, 2, 3]), np.array([2, 4, 6])
print(np.cov(X, Y))   # [[1,2],[2,4]]
````

**🧪 Try it myself:** Sign of `Cov` for `X=[1,2,3]`, `Y=[3,2,1]`? <details><summary>Solution</summary>Negative (they move oppositely).</details>

# Common Distributions

*📓 Notes · [03 · Probability](../README.md) · distributions*

| Name | Models | Formula | Example |
|---|---|---|---|
| Bernoulli(p) | one yes/no | `P(1)=p` | one coin |
| Binomial(n,p) | # of yeses in n tries | `C(n,k)pᵏ(1−p)ⁿ⁻ᵏ` | heads in 10 flips |
| Categorical | one of K classes | `P(k)=pₖ` | next word |
| Uniform | all equal | `1/(b−a)` | random init |
| Normal(μ,σ²) | bell curve | `1/(σ√2π) e^(−(x−μ)²/2σ²)` | noise, weights |

**✏️ Let me calculate it (binomial):** `P(2 heads in 3 flips) = 3 · 0.5² · 0.5 = 0.375`.

**🤖 Why AI uses it:** Bernoulli → binary classification; Categorical → softmax output; Normal → noise, weight init, Gaussian assumptions.

````python
from math import comb
print(comb(3, 2) * 0.5**3)                # 0.375
w = np.random.default_rng(0).normal(0, 0.1, size=5)   # weight init
print(w)
````

**🧪 Try it myself:** P(exactly 1 head in 2 flips)? <details><summary>Solution</summary>`2 · 0.5 · 0.5 = 0.5`.</details>

# Joint Distributions

*📓 Notes · [03 · Probability](../README.md) · distributions*

**📝 In simple words:** Probabilities of two variables together.

**🔢 Small example:** weather × umbrella:

| | umbrella | no umbrella | **marginal** |
|---|---|---|---|
| rain | 0.30 | 0.05 | **0.35** |
| sun | 0.10 | 0.55 | **0.65** |

**📐 Formula to remember:** marginal `P(X) = Σᵧ P(X,Y)`; conditional `P(Y|X) = P(X,Y)/P(X)`.

**✏️ Let me calculate it:** `P(umbrella | rain) = 0.30 / 0.35 ≈ 0.857`.

**🤖 Why AI uses it:** the probability of a whole sentence is a joint distribution over tokens, factored into conditionals (chain rule of probability) in chapter 10.

````python
J = np.array([[0.30, 0.05], [0.10, 0.55]])
print(J.sum(axis=1), J[0, 0] / J[0].sum())
````

**🧪 Try it myself:** `P(rain | umbrella)`? <details><summary>Solution</summary>`0.30 / (0.30+0.10) = 0.75`.</details>
