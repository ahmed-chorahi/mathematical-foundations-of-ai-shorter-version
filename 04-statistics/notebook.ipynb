# Mean, Median, Mode

*📓 Notes · [04 · Statistics](../README.md) · describing data*

**📝 In simple words:** Three ways to describe the "center" of data.

**🔢 Small example:** `[2, 4, 4, 6, 9]`

**📐 Formula to remember:** mean `= Σxᵢ/n`; median = middle value when sorted; mode = most frequent.

**✏️ Let me calculate it:** mean `25/5 = 5`; median `4`; mode `4`. One outlier (say 90) would drag the mean to 17.7 but leave the median at 4.

**🤖 Why AI uses it:** the mean is the basis of the loss average and normalization; the median resists outliers (MAE vs MSE, chapter 07).

```python
import numpy as np
from collections import Counter
d = [2, 4, 4, 6, 9]
print(np.mean(d), np.median(d), Counter(d).most_common(1)[0][0])
```

**🧪 Try it myself:** Mean and median of `[1, 2, 3, 100]`? <details><summary>Solution</summary>Mean 26.5, median 2.5.</details>

# Variance & Standard Deviation

*📓 Notes · [04 · Statistics](../README.md) · describing data*

**📝 In simple words:** The average squared distance from the mean, and its square root.

**📐 Formula to remember:** `s² = Σ(xᵢ − x̄)² / (n−1)`, `s = √s²` (sample version uses `n−1`).

**✏️ Let me calculate it:** deviations `[−3,−1,−1,1,4]` → squares `[9,1,1,1,16]` → sum 28 → `28/4 = 7` → `s ≈ 2.65`.

**🤖 Why AI uses it:** feature standardization `(x − μ)/σ`, LayerNorm, and weight-initialization scales.

```python
print(np.var(d, ddof=1), np.std(d, ddof=1))   # 7.0  2.645...
z = (np.array(d) - np.mean(d)) / np.std(d)
print(z.round(2))
```

**🧪 Try it myself:** Population variance of `[1,3]`? <details><summary>Solution</summary>mean 2, `(1+1)/2 = 1`.</details>

# Correlation

*📓 Notes · [04 · Statistics](../README.md) · describing data*

**📝 In simple words:** How strongly two variables move together, scaled to `[−1, 1]`.

**📐 Formula to remember:** `r = Cov(X,Y) / (σₓσᵧ)`

**🔢 Small example:** `X=[1,2,3]`, `Y=[2,4,7]` → `r ≈ 0.99` (almost a straight line).

**🤖 Why AI uses it:** feature selection (drop near-duplicate features), spotting leakage. **Correlation ≠ causation.**

```python
print(np.corrcoef([1, 2, 3], [2, 4, 7])[0, 1])
```

**🧪 Try it myself:** Correlation of `X=[1,2,3]`, `Y=[6,4,2]`? <details><summary>Solution</summary>`−1` (perfect negative line).</details>

# Sampling

*📓 Notes · [04 · Statistics](../README.md) · sampling and confidence*

**📝 In simple words:** Studying a small group to learn about a big one.

**💡 Think of it like this:** Averages of bigger samples wobble less. The standard error is `σ/√n`: 4× the data = half the noise.

**🤖 Why AI uses it:** a training set is a sample of the real world; a test set estimates future performance; mini-batches are samples (05).

```python
rng = np.random.default_rng(0)
pop = rng.normal(50, 10, 100_000)
for n in (10, 100, 1000):
    means = [rng.choice(pop, n).mean() for _ in range(500)]
    print(n, np.std(means).round(2))   # ≈ 10/√n
```

**🧪 Try it myself:** SE for σ=10, n=25? <details><summary>Solution</summary>`10/5 = 2`.</details>

# Confidence Intervals

*📓 Notes · [04 · Statistics](../README.md) · sampling and confidence*

**📝 In simple words:** A range likely to contain the true mean.

**📐 Formula to remember:** `x̄ ± 1.96 · s/√n` (about 95%).

**🔢 Small example:** `x̄ = 50`, `s = 10`, `n = 100`.

**✏️ Let me calculate it:** `SE = 1` → `50 ± 1.96` → `[48.04, 51.96]`.

**🤖 Why AI uses it:** "accuracy 91%" means little without error bars, especially on small test sets.

```python
m, s, n = 50, 10, 100
se = s / np.sqrt(n); print(m - 1.96*se, m + 1.96*se)
```

**🧪 Try it myself:** Same, but `n = 400`? <details><summary>Solution</summary>`SE=0.5` → `[49.02, 50.98]`.</details>

# Hypothesis Testing

*📓 Notes · [04 · Statistics](../README.md) · testing*

**📝 In simple words:** Asking "could this result be just luck?"

**✏️ Let me calculate it:** (1) null hypothesis H₀ ("nothing special") (2) compute a test statistic (3) see how surprising it is under H₀.

**🔢 Small example:** a coin lands heads 60 of 100 times. H₀: fair coin.

**📐 Formula to remember:** `z = (observed − expected) / SE`, with `SE = √(n p (1−p)) = 5`.

**✏️ Let me calculate it:** `z = (60 − 50)/5 = 2`.

**🤖 Why AI uses it:** comparing models, A/B tests, checking that an improvement isn't noise.

```python
z = (60 - 50) / np.sqrt(100 * 0.5 * 0.5); print(z)   # 2.0
```

**🧪 Try it myself:** 55 heads in 100 flips: z? <details><summary>Solution</summary>`5/5 = 1`, not surprising.</details>

# p-values

*📓 Notes · [04 · Statistics](../README.md) · testing*

**📝 In simple words:** The probability of a result at least this extreme *if H₀ were true*.

**🔢 Small example:** `z = 2` → two-sided `p ≈ 0.046`.

**Rule of thumb:** `p < 0.05` is "statistically significant", but it's **not** the probability that H₀ is true, and it says nothing about effect size.

```python
from math import erf, sqrt
p = 2 * (1 - 0.5 * (1 + erf(2 / sqrt(2)))); print(round(p, 4))   # 0.0455
```

**🧪 Try it myself:** Is `z = 1` significant at 0.05? <details><summary>Solution</summary>`p ≈ 0.32`, so no.</details>

# Statistical Intuition for ML

*📓 Notes · [04 · Statistics](../README.md) · ml intuition*

| Statistics idea | ML version |
|---|---|
| sample vs population | train/test split |
| standard error shrinks with n | more data → steadier estimates |
| overfitting to noise | memorizing training samples |
| standardization | feature scaling, BatchNorm/LayerNorm |
| confidence interval | error bars on accuracy |
| outliers | MSE vs MAE choice |

```python
# a model's measured accuracy on a small test set is noisy
n, acc = 50, 0.9
print(1.96 * np.sqrt(acc*(1-acc)/n))   # ±~0.083, so 90% could really be 82–98%
```

**🧪 Try it myself:** Does going from 50 to 200 test samples help? <details><summary>Solution</summary>Yes, error halves (`√4 = 2`).</details>
