# Linear Regression

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · regression*

**📝 In simple words:** Fit a line `ŷ = wx + b` to data.

**💡 Think of it like this:** Choose the line that sits closest to all points.

**🔢 Small example:** `x = [1,2,3]`, `y = [2,4,5]`.

**📐 Formula to remember:** `w = Σ(x−x̄)(y−ȳ) / Σ(x−x̄)²`, `b = ȳ − w x̄`

**✏️ Let me calculate it:** `x̄=2`, `ȳ=3.667`; numerator `3`, denominator `2` → `w = 1.5`, `b = 3.667 − 3 = 0.667`. Predictions: `[2.17, 3.67, 5.17]`.

**🤖 Why AI uses it:** the simplest model, and a neuron *is* linear regression plus an activation.

````python
import numpy as np
x = np.array([1., 2., 3.]); y = np.array([2., 4., 5.])
w = np.sum((x - x.mean()) * (y - y.mean())) / np.sum((x - x.mean())**2)
b = y.mean() - w * x.mean()
print(w, b)                        # 1.5  0.667
# library equivalent: np.polyfit(x, y, 1)  or sklearn LinearRegression
print(np.polyfit(x, y, 1))
````

**🧪 Try it myself:** Predict at `x = 4`. <details><summary>Solution</summary>`1.5·4 + 0.667 = 6.667`.</details>

# MSE & MAE

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · regression*

**📝 In simple words:** Two ways to average the errors.

**📐 Formula to remember:** `MSE = (1/n) Σ(y − ŷ)²`, `MAE = (1/n) Σ|y − ŷ|`

**🔢 Small example:** `y = [3,5,7]`, `ŷ = [2,5,10]` → errors `[−1, 0, 3]`.

**✏️ Let me calculate it:** MSE `= (1+0+9)/3 = 3.33`; MAE `= (1+0+3)/3 = 1.33`. MSE punishes the big miss (3) much more.

**🤖 Why AI uses it:** MSE is smooth and standard for regression; MAE is robust to outliers.

````python
y = np.array([3, 5, 7]); yh = np.array([2, 5, 10])
print(np.mean((y - yh)**2), np.mean(np.abs(y - yh)))
````

**🧪 Try it myself:** MSE for `y=[1,2]`, `ŷ=[2,4]`? <details><summary>Solution</summary>`(1+4)/2 = 2.5`.</details>

# Logistic Regression & the Sigmoid

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · classification*

**📝 In simple words:** Linear regression squashed into a probability for yes/no.

**📐 Formula to remember:** `p = σ(z) = 1/(1+e⁻ᶻ)`, with `z = w·x + b`

**🔢 Small example:** `w = 2`, `b = −1`, `x = 1`.

**✏️ Let me calculate it:** `z = 1` → `p = 1/(1+e⁻¹) = 0.731`. `z = 0` → 0.5; large `z` → near 1; very negative → near 0.

**🤖 Why AI uses it:** binary classification, and the sigmoid is a classic activation/gate.

````python
sigmoid = lambda z: 1 / (1 + np.exp(-z))
print(sigmoid(np.array([-4, 0, 1, 4])).round(3))   # [0.018 0.5 0.731 0.982]
````

**🧪 Try it myself:** `σ(0)`? <details><summary>Solution</summary>0.5.</details>

# Binary Cross-Entropy

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · classification*

**📝 In simple words:** The loss for yes/no predictions (chapter 06's cross-entropy with 2 classes).

**📐 Formula to remember:** `BCE = −[y ln p + (1−y) ln(1−p)]`

**🔢 Small example:** `y = 1`.

**✏️ Let me calculate it:** `p = 0.9` → `−ln 0.9 = 0.105`. `p = 0.1` → `2.303`. For `y=[1,0]`, `p=[0.9,0.2]`: `(0.105 + 0.223)/2 = 0.164`.

**🤖 Why AI uses it:** every binary classifier, plus multi-label outputs.

````python
y = np.array([1, 0]); p = np.array([0.9, 0.2])
print(-np.mean(y*np.log(p) + (1-y)*np.log(1-p)))   # 0.164
````

**🧪 Try it myself:** `y=0`, `p=0.5`? <details><summary>Solution</summary>`−ln 0.5 = 0.693`.</details>

# Softmax

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · classification*

**📝 In simple words:** Turns any list of scores (logits) into probabilities.

**📐 Formula to remember:** `softmax(z)ᵢ = e^zᵢ / Σⱼ e^zⱼ`

**🔢 Small example:** `z = [2, 1, 0]`.

**✏️ Let me calculate it:** `e² = 7.389`, `e¹ = 2.718`, `e⁰ = 1` → sum `11.107` → `[0.665, 0.245, 0.090]`. All positive, sum to 1, bigger score → bigger share.

**Stability trick:** subtract `max(z)` first; the answer is unchanged but `e^1000` won't overflow.

**🤖 Why AI uses it:** multi-class output, attention weights (09), next-token probabilities (10).

````python
def softmax(z):
    e = np.exp(z - np.max(z))
    return e / e.sum()
print(softmax(np.array([2., 1., 0.])).round(3))   # [0.665 0.245 0.09]
````

**🧪 Try it myself:** `softmax([0, 0])`? <details><summary>Solution</summary>`[0.5, 0.5]`.</details>

# Maximum Likelihood

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · likelihood and regularization*

**📝 In simple words:** Choose parameters that make the observed data *most probable*.

**🔢 Small example:** a coin gave 7 heads out of 10. Which `p` fits best?

**📐 Formula to remember:** `L(p) = p⁷(1−p)³`; log-likelihood `ℓ = 7 ln p + 3 ln(1−p)`

**✏️ Let me calculate it:** `dℓ/dp = 7/p − 3/(1−p) = 0` → `p = 0.7`.

**The key link:** *minimizing cross-entropy = maximizing log-likelihood*. MSE is MLE under Gaussian noise; BCE is MLE under Bernoulli labels.

**🤖 Why AI uses it:** it's the principled reason behind our loss functions, including the LLM objective.

````python
ps = np.linspace(0.01, 0.99, 99)
ll = 7*np.log(ps) + 3*np.log(1-ps)
print(ps[np.argmax(ll)])   # 0.7
````

**🧪 Try it myself:** 3 heads in 4 flips: MLE of `p`? <details><summary>Solution</summary>`3/4 = 0.75`.</details>

# Regularization

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · likelihood and regularization*

**📝 In simple words:** Add a penalty for large weights to stop overfitting.

**📐 Formula to remember:** `L_total = L_data + λ‖w‖²` (L2/ridge) or `λ Σ|wᵢ|` (L1/lasso)

**🔢 Small example:** `w = [3, 4]`, `λ = 0.1`.

**✏️ Let me calculate it:** L2 penalty `= 0.1·(9+16) = 2.5`. The gradient adds `2λw`, so every update also shrinks `w` a bit ("weight decay"). L1 tends to push weights to exactly 0 (sparse).

**🤖 Why AI uses it:** weight decay is standard in training Transformers (AdamW).

````python
w = np.array([3., 4.]); lam = 0.1
print(lam * np.sum(w**2), lam * np.sum(np.abs(w)))   # 2.5  0.7
````

**🧪 Try it myself:** L2 penalty for `w=[1,1,1]`, `λ=0.5`? <details><summary>Solution</summary>`0.5·3 = 1.5`.</details>

# PCA (as an ML tool)

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · pca and generalization*

**📝 In simple words:** Linear-algebra PCA from chapter 01, used to shrink features.

**✏️ Let me calculate it:** center → SVD → keep top-k directions → project. "Explained variance" `= σᵢ² / Σσ²`.

**🤖 Why AI uses it:** speed up training, visualize embeddings in 2D, denoise.

````python
rng = np.random.default_rng(1)
X = rng.normal(size=(100, 3)) @ np.array([[3, 0, 0], [0, 1, 0], [0, 0, 0.1]])
Xc = X - X.mean(0)
U, S, Vt = np.linalg.svd(Xc, full_matrices=False)
print((S**2 / np.sum(S**2)).round(3))   # first component dominates
Z = Xc @ Vt[:2].T                       # 3D → 2D
````

**🧪 Try it myself:** If `S² = [90, 9, 1]`, how many components keep ≥ 95%? <details><summary>Solution</summary>Two (`99%`). One alone is 90%.</details>

# Bias–Variance Intuition

*📓 Notes · [07 · Machine Learning Mathematics](../README.md) · pca and generalization*

**📝 In simple words:** Error = bias² + variance + noise.
- **High bias** (too simple) → underfits: misses the pattern.
- **High variance** (too flexible) → overfits: memorizes noise.

```text
 error
   │\                 /   test
   │ \               /
   │  \___________  /
   │   train   ↘ __/
   └─────────────────► model complexity
    underfit   sweet spot   overfit
```

**🤖 Why AI uses it:** explains train/test gaps; regularization, more data, and early stopping all fight variance.

````python
rng = np.random.default_rng(0)
f = lambda x: np.sin(2*np.pi*x)
xt = rng.uniform(0, 1, 20); yt = f(xt) + rng.normal(0, 0.2, 20)
xs = rng.uniform(0, 1, 200); ys = f(xs) + rng.normal(0, 0.2, 200)
for deg in (1, 3, 9):
    c = np.polyfit(xt, yt, deg)
    print(deg, np.mean((np.polyval(c, xt)-yt)**2).round(3), np.mean((np.polyval(c, xs)-ys)**2).round(3))
# degree 1: both high | 3: both low | 9: train tiny, test larger
````

**🧪 Try it myself:** Train error 0.01, test error 0.9: bias or variance problem? <details><summary>Solution</summary>Variance (overfitting).</details>
