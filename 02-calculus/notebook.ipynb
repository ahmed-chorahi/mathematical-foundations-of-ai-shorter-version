# Functions & Limits

*📓 Notes · [02 · Calculus](../README.md) · limits and derivatives*

**📝 In simple words:** A limit is the value a function approaches as `x` gets close to a point.

**💡 Think of it like this:** Zoom in until you can't tell the difference.

**🔢 Small example:** `f(x) = (x² − 1)/(x − 1)` at `x → 1`. Can't plug in 1 (0/0), but it simplifies.

**📐 Formula to remember:** `lim_{x→a} f(x)`

**✏️ Let me calculate it:** `(x−1)(x+1)/(x−1) = x+1` → limit is `2`.

**🤖 Why AI uses it:** a derivative *is* a limit; it's the foundation under everything below.

````python
f = lambda x: (x**2 - 1) / (x - 1)
for x in [1.1, 1.01, 1.001]: print(x, f(x))   # → 2
````

**🧪 Try it myself:** `lim_{x→2} (x²−4)/(x−2)`? <details><summary>Solution</summary>`x+2 → 4`.</details>

# Derivatives

*📓 Notes · [02 · Calculus](../README.md) · limits and derivatives*

**📝 In simple words:** The slope of a function at a point.

**💡 Think of it like this:** Speedometer reading: how fast the output changes right now.

**🔢 Small example:** `f(x) = x²` at `x = 3`.

**📐 Formula to remember:** `f′(x) = lim_{h→0} (f(x+h) − f(x)) / h`. Rules: `(xⁿ)′ = n xⁿ⁻¹`, `(eˣ)′ = eˣ`, `(ln x)′ = 1/x`.

**✏️ Let me calculate it:** `f′(x) = 2x` → `f′(3) = 6`. Check: `(3.001² − 3²)/0.001 ≈ 6.001`.

**🤖 Why AI uses it:** the slope of the loss tells us which way to move a weight.

````python
f = lambda x: x**2
h = 1e-5
print((f(3 + h) - f(3)) / h)   # ≈ 6
````

**🧪 Try it myself:** Derivative of `x³` at `x=2`? <details><summary>Solution</summary>`3x² = 12`.</details>

# Partial Derivatives

*📓 Notes · [02 · Calculus](../README.md) · many variables*

**📝 In simple words:** The derivative with respect to one variable, treating the others as constants.

**🔢 Small example:** `f(x, y) = x²y`

**📐 Formula to remember:** `∂f/∂x`, `∂f/∂y`

**✏️ Let me calculate it:** `∂f/∂x = 2xy`, `∂f/∂y = x²`. At `(2,3)`: `12` and `4`.

**🤖 Why AI uses it:** a model has millions of weights; each one gets its own partial derivative of the loss.

````python
f = lambda x, y: x**2 * y
h = 1e-5
print((f(2+h, 3) - f(2, 3)) / h, (f(2, 3+h) - f(2, 3)) / h)   # ≈ 12, 4
````

**🧪 Try it myself:** `∂/∂y` of `3xy + y²` at `(1,2)`? <details><summary>Solution</summary>`3x + 2y = 3 + 4 = 7`.</details>

# Chain Rule

*📓 Notes · [02 · Calculus](../README.md) · many variables*

**📝 In simple words:** The derivative of a function inside a function.

**💡 Think of it like this:** Effects multiply along a chain: if A changes B by 2× and B changes C by 3×, A changes C by 6×.

**🔢 Small example:** `y = (3x + 1)²`

**📐 Formula to remember:** `dy/dx = dy/du · du/dx`

**✏️ Let me calculate it:** `u = 3x+1`: `dy/du = 2u`, `du/dx = 3` → `6(3x+1)`. At `x=1`: `24`.

**🤖 Why AI uses it:** a network is layers inside layers. Backpropagation *is* the chain rule applied layer by layer.

````python
f = lambda x: (3*x + 1)**2
h = 1e-5
print((f(1 + h) - f(1)) / h)   # ≈ 24
````

**🧪 Try it myself:** Derivative of `(2x)³` at `x=1`? <details><summary>Solution</summary>`3(2x)²·2 = 3·4·2 = 24`.</details>

# Gradients

*📓 Notes · [02 · Calculus](../README.md) · many variables*

**📝 In simple words:** The vector of all partial derivatives.

**💡 Think of it like this:** It points uphill, the direction of fastest increase. Go the opposite way to go downhill.

**🔢 Small example:** `f(x,y) = x² + y²` at `(1, 2)`.

**📐 Formula to remember:** `∇f = [∂f/∂x, ∂f/∂y]`

**✏️ Let me calculate it:** `∇f = [2x, 2y] = [2, 4]`. Downhill direction: `[−2, −4]`.

**🤖 Why AI uses it:** `∇Loss` tells every weight which way to move. Core of gradient descent.

````python
import numpy as np
grad = lambda p: 2 * p            # for f = x² + y²
print(grad(np.array([1., 2.])))   # [2. 4.]
````

**🧪 Try it myself:** Gradient of `x² + 3y` at `(2,5)`? <details><summary>Solution</summary>`[2x, 3] = [4, 3]`.</details>

# Jacobian

*📓 Notes · [02 · Calculus](../README.md) · many variables*

**📝 In simple words:** The matrix of derivatives for a function with many inputs *and* many outputs.

**🔢 Small example:** `f(x,y) = [x+y, xy]`

**📐 Formula to remember:** `J_ij = ∂fᵢ/∂xⱼ`

**✏️ Let me calculate it:** `J = [[1, 1],[y, x]]`. At `(2,3)`: `[[1,1],[3,2]]`.

**🤖 Why AI uses it:** a layer maps a vector to a vector; backprop multiplies Jacobians (e.g., the softmax Jacobian).

````python
J = lambda x, y: np.array([[1, 1], [y, x]])
print(J(2, 3))
````

**🧪 Try it myself:** Jacobian of `[2x, x+3y]`? <details><summary>Solution</summary>`[[2,0],[1,3]]`.</details>

# Hessian

*📓 Notes · [02 · Calculus](../README.md) · many variables*

**📝 In simple words:** The matrix of *second* derivatives, describing the curvature.

**🔢 Small example:** `f(x,y) = x² + 3y²`

**📐 Formula to remember:** `H_ij = ∂²f/∂xᵢ∂xⱼ`

**✏️ Let me calculate it:** `H = [[2, 0],[0, 6]]`. Both positive → bowl shape → minimum.

**🤖 Why AI uses it:** it tells whether a flat spot is a min, max, or saddle; second-order optimizers use it, and it explains why some directions are harder to learn.

````python
H = np.array([[2, 0], [0, 6]])
print(np.linalg.eigvalsh(H))   # all > 0 → minimum
````

**🧪 Try it myself:** `f = x² − y²`. Min, max, or saddle? <details><summary>Solution</summary>`H = diag(2, −2)`, mixed signs → saddle.</details>

# Critical Points

*📓 Notes · [02 · Calculus](../README.md) · critical points*

**📝 In simple words:** Places where the gradient is zero.

**🔢 Small example:** `f(x) = x² − 4x + 5`

**📐 Formula to remember:** `f′(x) = 0`; `f″ > 0` → min, `f″ < 0` → max.

**✏️ Let me calculate it:** `f′ = 2x − 4 = 0` → `x = 2`; `f″ = 2 > 0` → minimum, `f(2) = 1`.

**🤖 Why AI uses it:** training aims for a point where the gradient is ~0 (a minimum of loss).

````python
f = lambda x: x**2 - 4*x + 5
xs = np.linspace(-1, 5, 7); print(xs[np.argmin(f(xs))])   # 2.0
````

**🧪 Try it myself:** Critical point of `x² + 6x`? <details><summary>Solution</summary>`2x + 6 = 0` → `x = −3`.</details>

# Integrals

*📓 Notes · [02 · Calculus](../README.md) · integrals*

**📝 In simple words:** The area under a curve; the sum of tiny slices.

**🔢 Small example:** area under `f(x) = x` from 0 to 2.

**📐 Formula to remember:** `∫ₐᵇ f(x)dx ≈ Σ f(xᵢ)Δx`

**✏️ Let me calculate it:** `∫₀² x dx = x²/2 |₀² = 2`.

**🤖 Why AI uses it:** probabilities of continuous variables are areas; expected values are integrals.

````python
x = np.linspace(0, 2, 1001)
print(np.sum(x[:-1] * np.diff(x)))   # ≈ 2
````

**🧪 Try it myself:** `∫₀³ 2x dx`? <details><summary>Solution</summary>`x² |₀³ = 9`.</details>

# Why Derivatives Matter in ML

*📓 Notes · [02 · Calculus](../README.md) · derivatives in ml*

```text
 loss
  │\
  │ \
  │  \__          slope negative → move right
  │     \__/      slope zero     → stop (minimum)
  └──────────► weight
```

1. Define a **loss** (how wrong the model is).
2. Compute the **derivative** of loss with respect to each weight.
3. Move each weight a little **against** its derivative.
4. Repeat. That's learning.

````python
# learn w so that w*2 ≈ 6  (true w = 3)
w, lr = 0.0, 0.1
for _ in range(50):
    grad = 2 * (w*2 - 6) * 2     # d/dw of (2w − 6)²
    w -= lr * grad
print(round(w, 3))               # ≈ 3.0
````

**🧪 Try it myself:** In the code above, what's the loss derivative at `w=0`? <details><summary>Solution</summary>`2(−6)(2) = −24`: negative, so `w` increases.</details>
